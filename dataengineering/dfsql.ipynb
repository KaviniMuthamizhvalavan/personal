```python
import pyspark

import os
import sys
from pyspark.sql import SparkSession

# Dynamically point PySpark to YOUR current active Python environment
os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

# Network binding fix for local execution
os.environ["PYSPARK_SUBMIT_ARGS"] = "--conf spark.driver.host=127.0.0.1 --conf spark.driver.bindAddress=127.0.0.1 pyspark-shell"

# Initialize Spark Session
spark = SparkSession.builder \
    .appName("RDD Impl") \
    .master("local[2]") \
    .config("spark.driver.host", "127.0.0.1") \
    .config("spark.driver.bindAddress", "127.0.0.1") \
    .getOrCreate()


df = spark.read.csv(
    "employees_records.csv",
    header=True,
    inferSchema=True
)

df.show()
df.printSchema()
```

```python
df.filter(
    df.salary > 70000
).show()
```

```python
df.createOrReplaceTempView("employee")
spark.sql("""select * from employee where salary > 70000""").show()
```

```python
df.filter(
    df.department == "IT"
).show()
```

```python
spark.sql("""select * from employee where department = 'IT'""").show()
```

```python
spark.sql("""select AVG(salary) from employee""").show()
```

```python
spark.sql("""select MAX(salary) from employee""").show()
```

```python
spark.sql("""select department, COUNT(*) as employee_count from employee group by department""").show()
```
